# Rainfall summary

A short look at a month of rainfall readings.


```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(20260819)
```

```python
readings = rng.gamma(shape=2.0, scale=3.0, size=31).round(1)
readings[:7]
```

```python
total = readings.sum()
wettest = readings.max()
print(f'total {total:.1f} mm')
print(f'wettest day {wettest:.1f} mm')
```

```python
days = np.arange(1, len(readings) + 1)
rolling_average = np.convolve(readings, np.ones(5) / 5, mode='valid')
fig, ax = plt.subplots()
ax.bar(days, readings, alpha=0.35, label='daily')
ax.plot(days[2:-2], rolling_average, color='tab:orange', label='5-day average')
ax.set_xlabel('day')
ax.set_ylabel('rainfall (mm)')
ax.legend()
plt.show()
```

## Notes

Readings are provisional until the station is recalibrated.
